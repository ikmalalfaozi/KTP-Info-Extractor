```python
from ultralytics import YOLO

# Load a model
model = YOLO("yolo11n-cls.pt")  # load a pretrained model (recommended for training)

# Train the model
results = model.train(data="/kaggle/input/ktp-orientation-classification", epochs=50, imgsz=640)
```

```python
# Evaluate model on validation or val set
metrics = model.val(data="/kaggle/input/ktp-orientation-classification", split='val')
print(metrics)
```

```python
# Evaluate model on validation or test set
metrics = model.val(data="/kaggle/input/ktp-orientation-classification", split='test')
print(metrics)
```

```python
from ultralytics import YOLO
import matplotlib.pyplot as plt
import cv2
import numpy as np

# Path to image for inference
image_path = "/kaggle/input/ktp-orientation-classification/test/270/100_jpg.rf.b692baacb3b5387d15da76f4b0068b82_ktp_0.jpg"

# Make predictions
results = model.predict(image_path)
probs = results[0].probs.data.tolist()
names_dict = results[0].names

# Taking predictions with maximum probability
max_prob = max(probs)
max_class = names_dict[np.argmax(probs)]

# show images
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

label = f"{max_class}: {max_prob:.2f}"
font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(image, label, (10, 30), font, 1, (255, 0, 0), 2, cv2.LINE_AA)

plt.figure(figsize=(8, 6))
plt.imshow(image)
plt.axis("off")
plt.title(f"Predicted Class: {max_class}")
plt.show()

print(f"Predicted Class: {max_class}, Probability: {max_prob:.4f}")
```
